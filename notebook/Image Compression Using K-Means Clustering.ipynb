```python
import numpy as np

import pandas as pd

from PIL import Image

from sklearn.cluster import KMeans

import matplotlib.pyplot as plt

```

```python
# Load image
```

```python
image = Image.open('../dataset/parrot.jpg').convert("RGB")
```

```python
#convert image to numpy array
```

```python
img_array = np.array(image)
```

```python
img_array
```

```python
# Get image dimesions
```

```python
height , width , channels = img_array.shape
```

```python
height , width , channels 
```

```python
# convert picture into RGB rows
```

```python
pixels = img_array.reshape(-1,3)
```

```python
pixels
```

```python
# Number of colors
```

```python
k = 16
```

```python
# Kmeans Clustering
```

```python
kmeans = KMeans(
    n_clusters=k,
    random_state=42,
    n_init=10
)


kmeans.fit(pixels)
```

```python
# clusters center
centers = np.uint8(kmeans.cluster_centers_)
```

```python
centers
```

```python
# Replace the pixel with cluster center 
```

```python
compressed_pixel = centers[kmeans.labels_]
```

```python
compressed_pixel
```

```python
# Reshape back to image
```

```python
compressed_image = compressed_pixel.reshape(height, width, 3)
```

```python
compressed_image
```

```python
# save Compressed image
```

```python
Image.fromarray(compressed_image).save('compressed.jpg',quality=70)
```

```python
# Display
plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(image)
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(compressed_image)
plt.title(f"Compressed Image - K={k}")
plt.axis("off")

plt.show()
```
